```python
# Configures matplotlib to show figures embedded in the notebook 
%matplotlib inline

from pylab import *
import matplotlib 
import matplotlib.pyplot as plt
import numpy as np
import materials as materials
```

```python
#quickly checking the materials details in the table
dir(materials)
```

```python
# Temperature range from 294 K to 600 K
T = np.linspace(294, 600, 200)
```

```python
#we should define simple functions for each material
```

```python
# Functions for density evaluation
def density_mercury(T):
    return materials.mercury.density(T)

def density_graphite(T):
    return materials.graphite.density(T)

def density_copper(T):
    return materials.copper.density(T)

def density_water(T):
    return materials.water.density(T)
```

```python
#We evaluate the density functions over the temperature range.
```

```python
#Compute densities
rho_mercury  = density_mercury(T)
rho_graphite = density_graphite(T)
rho_copper   = density_copper(T)
rho_water    = density_water(T)
```

```python
#We plot all four materials in a single figure.
```

```python
plt.figure(figsize=(10,6))

plt.plot(T, rho_mercury,  label="Mercury")
plt.plot(T, rho_graphite, label="Graphite")
plt.plot(T, rho_copper,   label="Copper")
plt.plot(T, rho_water,    label="Water")

plt.xlabel("Temperature (K)")
plt.ylabel("Density (kg/m³)")
plt.title("Temperature-dependent Density of Materials")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#We create a 2×2 grid of subplots, one for each material.
```

```python
#Separate subplots
fig, axs = plt.subplots(2, 2, figsize=(12,10))
axs = axs.flatten()

names = ["Mercury", "Graphite", "Copper", "Water"]
rhos  = [rho_mercury, rho_graphite, rho_copper, rho_water]

for i in range(4):
    axs[i].plot(T, rhos[i])
    axs[i].set_title(names[i])
    axs[i].set_xlabel("Temperature (K)")
    axs[i].set_ylabel("Density (kg/m³)")
    axs[i].grid(True)

plt.tight_layout()
plt.show()
```
