## Questions - GDP Per Person (1901-2011)

```python
df_gdp = pd.read_csv("Datasets/GDP.csv", decimal=".")
df_gdp.head()
#df_gdp.columns
```

- Clean the dataset by converting strings to dates or floats where necessary.

```python
df_gdp.columns = df_gdp.columns.str.strip()
df_gdp['Year'] = pd.to_datetime(df_gdp['Year'], errors='coerce')
df_gdp['Year'] = df_gdp['Year'].dt.year
df_gdp['GDP_pp'] = pd.to_numeric(df_gdp['GDP_pp'], errors='coerce')
df_gdp_clean = df_gdp.dropna(subset=['Year', 'GDP_pp'])
df_gdp_clean['Year'] = df_gdp_clean['Year'].astype(int)
df_gdp_clean.head()
```

- Can you report the first recorded value for each country?

```python
df_first = (
    df_gdp_clean
    .sort_values('Year')
    .groupby('Country')
    .first()
    .reset_index()
)

df_first[['Country', 'Year', 'GDP_pp']]
```

- Identify the regions with the largest increases in GDP per capita during the twentieth century.

```python
# Restrict to 20th century: 1901–2001
df_20th = df_gdp_clean[(df_gdp_clean['Year'] >= 1901) & (df_gdp_clean['Year'] <= 2001)]

# country-level increase within 20th century
country_change_20th = (
    df_20th
    .sort_values('Year')
    .groupby('Country')
    .agg(
        first_gdp=('GDP_pp', 'first'),
        last_gdp=('GDP_pp', 'last'),
        region=('Region', 'first')
    )
)

country_change_20th['Increase'] = country_change_20th['last_gdp'] - country_change_20th['first_gdp']

#  top country per region, sorted by increase
region_top_country_20th = (
    country_change_20th
    .sort_values('Increase', ascending=False)
    .groupby('region')
    .head(1)
    .sort_values('Increase', ascending=False)
)

region_top_country_20th
```

- Fill in the missing years for each country with estimates based on the difference between the next and previous records.

```python
all_years = range(df_gdp_clean['Year'].min(),
                  df_gdp_clean['Year'].max() + 1)

filled_gdp = (
    df_gdp_clean
    .set_index('Year')
    .groupby('Country')['GDP_pp']
    .apply(lambda s: s.reindex(all_years).interpolate())
    .reset_index()
)

filled_gdp.head()
```
