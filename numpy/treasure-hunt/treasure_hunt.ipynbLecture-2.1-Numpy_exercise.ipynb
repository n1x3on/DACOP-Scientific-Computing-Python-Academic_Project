7. **Treasure Hunt**

Develop a small game where the goal is to find a treasure hidden on a map. The map is a two-dimensional 5x5 `NumPy` array in which each cell represents a position. The treasure is represented by a random number in the two-dimensional `NumPy` array, and the player must find it as quickly as possible. The player can move in four directions: up, down, left, or right.

##### Rules

* The player always starts at position `(0, 0)` on the map.
* The player can move only to a neighboring cell abd they **cannot** skip cells.
* The game ends when the treasure is found.
* The player receives a score based on the number of moves needed to find the treasure.

```python
import numpy as np
import random

# Create 5x5 map
map_ = np.zeros((5, 5), dtype=int)

# Random treasure position
treasure_pos = tuple(np.random.randint(0, 5, size=2))
map_[treasure_pos] = 1  # mark treasure internally

# Player starts at (0,0)
player_pos = [0, 0]
moves = 0
#print(map_)
print("Treasure Hunt Game Started!")
print("You are at position (0,0). Move using: up, down, left, right")

while True:
    print(f"\nCurrent position: {tuple(player_pos)}")
    move = input("Enter move: ").strip().lower()

    if move == "w" and player_pos[0] > 0:
        player_pos[0] -= 1
    elif move == "s" and player_pos[0] < 4:
        player_pos[0] += 1
    elif move == "a" and player_pos[1] > 0:
        player_pos[1] -= 1
    elif move == "d" and player_pos[1] < 4:
        player_pos[1] += 1
    else:
        print("Invalid move. Try again.")
        continue

    moves += 1

    # Check treasure
    if tuple(player_pos) == treasure_pos:
        print("\n Treasure found!")
        print(f"Treasure was at: {treasure_pos}")
        print(f"Total moves: {moves}")
        print(map_)
        break
```
