### OOP Exercise 1 : Gym Chaos Simulator

Consider a gym with a dumbbell rack. Each dumbbell has a correct predefined position on the rack. For example, the 12 kg dumbbell belongs in the 12 kg position, the 14 kg dumbbell in the 14 kg position, and so on.

There are two types of gym users: organized users and chaotic users. The simulation should model how their behavior affects the organization of the rack over time.

The behavior is governed by the following three rules:

- **Organized user:**
    - Correct position available: After using a dumbbell, an organized user checks its correct position. If that position is free, the user returns the dumbbell to its correct place.
    - Correct position occupied: If the correct position is already occupied by another dumbbell, the organized user does not reorganize the rack. Instead, the dumbbell is placed in another available position.

- **Chaotic user:**
    - A chaotic user does not care about the correct position of the dumbbell. After using it, the user places it in a random available position, even if its correct position is free.

The objective of the simulation is to investigate how a small proportion of chaotic users can affect the overall organization of a shared environment, even when the majority of users behave in an organized manner. For example, you could compare simulations such as 1 chaotic user among 10, 20, or 30 organized users and observe how the organization of the rack evolves over time.

```python
import random

class Gym:
    def __init__(self):
        self.dumbbell = [i for i in range(10, 36) if i % 2 == 0]
        self.rack = {}
        self.restart_day()

    # A function/method to restart the rack everyday
    def restart_day(self):
        self.rack = {i: i for i in self.dumbbell}

    # A function/method that gives the actual occupation status of the rack
    def print_dumbbells(self):
        return [i for i in self.rack.values() if i != 0]

    # A function/method that lists the places
    def print_places(self):
        return [i for i,j in self.rack.items() if j == 0]

    # A function/method that describes the dynamics of when an user gets a weight from the dumbbell rack
    def get_dumbbell(self, weight):
        dumbbell_pos = list(self.rack.values()).index(weight)                       # Get the index where a certain weight is
        key_dumbbell = list(self.rack.keys())[dumbbell_pos]                         # Get the rack label for that position 
        self.rack[key_dumbbell] = 0
        return weight

    # A function/method that describes the dynamics of when an user retuens a weight to the dumbbell rack
    def return_dumbbell(self, pos, weight):
        self.rack[pos] = weight

    # A function/method to analyse the current caos status
    def compute_caos(self):
        num_caos = [i for i, j in self.rack.items() if i != j]
        return (len(num_caos) / len(self.rack)) * 100
        

class Users:
    def __init__(self, user_type, gym):
        self.user_type = user_type            # 1 - normal , 2- unnorganized 
        self.gym = gym                        # call by reference - all users are pointing to this gym (if gym given as the argument is always the same)
        self.weight = 0                       # no weight

    def begin_training(self):
        weight_list = self.gym.print_dumbbells()            # List all the weights available in the rack
        self.weight = random.choice(weight_list)            # Select a weight at random
        self.gym.get_dumbbell(self.weight)

    def finish_training(self):
        dumbbell_places = self.gym.print_places()
        if self.user_type == 1:
            if self.weight in dumbbell_places:
                self.gym.return_dumbbell(self.weight, self.weight)
            else:
                pos = random.choice(dumbbell_places)
                self.gym.return_dumbbell(pos, self.weight)

        if self.user_type == 2:
                # Always randomically return the dumbbell to the dumbbell hack
                pos = random.choice(dumbbell_places)
                self.gym.return_dumbbell(pos, self.weight)

        # After returning the weight, the user does not have any weight with him/her anymore.
        self.weight = 0
                
                
# Lets create a gym instance                
gym = Gym()

# Create 10 normal users
users = [Users(1, gym) for i in range(10)]
users += [Users(2, gym) for i in range(1)]

# Lets see our list of users
for i in users:
    print(i.user_type)

# Lets randomize our list of users
random.shuffle(users)

# Lets make them go through 10 training sessions:
gym.restart_day()
for i in range(10):
    random.shuffle(users)
    for user in users:
        user.begin_training()
    for user in users:
        user.finish_training()

# Lets take a look at our Rack at the end of the gym sessions:
gym.rack
gym.compute_caos()


# This is like a Monte-Carlo simulation, where we developed a mathematical model about some problem, compute it and simulate it many times to get the mean result
# Lets get that mean outcome:
list_chaos = []

for k in range(2000):
    gym.restart_day()
    for i in range(10):
        random.shuffle(users)
        for user in users:
            user.begin_training()
        for user in users:
            user.finish_training()
        list_chaos += [gym.compute_caos()]

import matplotlib.pyplot as plt
plt.hist(list_chaos)
```
