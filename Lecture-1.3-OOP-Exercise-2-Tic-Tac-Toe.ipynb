### OOP Exercise 2: Tic-Tac-Toe

Consider a game of tic-tac-toe, played on a board with three rows and three columns. Develop a text-based Python game in which a human player competes against the computer. The human player uses **X**, the computer uses **O**, and each game begins with an empty board.

The game is governed by the following rules:

- **Human player:**
    - The human player always makes the first move. On each turn, the player chooses a position by entering its row and column, both numbered from 0 to 2.
    - A move is valid only if the chosen position is inside the board and currently empty. An existing mark cannot be overwritten.
    - If the input is invalid, explain the problem and ask the player to try again. This includes non-integer input, coordinates outside the board, and positions that are already occupied. Invalid input must not change the board or pass the turn to the computer.

- **Computer opponent:**
    - After each valid human move, provided the game has not ended, the computer places exactly one O in a randomly selected empty position.
    - The human player and the computer continue taking turns until the game ends.

- **Winning and drawing:**
    - The first player to place three of their marks in the same row, column, or diagonal wins.
    - If all nine positions are occupied and neither player has won, the game ends in a draw.
    - The game ends immediately after a winning move or a draw. No further moves are allowed in that game.

Display the empty board before the first move and the updated board after every valid move. Make the row and column labels clear so that the player can identify each position. When the game ends, display the final result and ask whether the player wants to play again or exit. A new game must begin with an empty board and the human player's turn.

The objective of this exercise is to design and implement a complete interactive game using **object-oriented programming**. Decide independently how to organize your solution.

**Example of a completed game:** X wins along the diagonal from the top right to the bottom left. A dash indicates an empty position.

| Row / Column | 0 | 1 | 2 |
|:------------:|:-:|:-:|:-:|
| **0** | X | O | **X** |
| **1** | O | **X** | — |
| **2** | **X** | — | O |


#### Solution

```python
import random

class tictactoe:
 def __init__(self):
     self.board = [["-" for _ in range(3)] for _ in range(3)]
 
 def print_board(self):
    print("Row / Column   0   1   2")
    for r in range(3):
        print(f"{r:<14}{self.board[r][0]:^3}|{self.board[r][1]:^3}|{self.board[r][2]:^3}")
        if r < 2:
            print("               ---------")

 def is_valid_move(self, r, c):
    if not (0 <= r < 3 and 0 <= c < 3):
        return False
    return self.board[r][c] == "-"

 def make_human_move(self):
    while True:
        try:
            r = int(input("Enter row (0-2): "))
            c = int(input("Enter column (0-2): "))
        except ValueError:
            print("Invalid input: please enter integers.")
            continue

        if self.is_valid_move(r, c):
            self.board[r][c] = "X"
            return True
        else:
            print("Invalid move: position is outside board or already occupied.")

 
 def make_computer_move(self):
    empty_positions = [(r, c) for r in range(3) for c in range(3) if self.board[r][c] == "-"]
    if empty_positions:
        r, c = random.choice(empty_positions)
        self.board[r][c] = "O"
         
 def check_winner(self):
    b = self.board

    # rows
    for r in range(3):
        if b[r][0] == b[r][1] == b[r][2] != "-":
            return b[r][0]

    # columns
    for c in range(3):
        if b[0][c] == b[1][c] == b[2][c] != "-":
            return b[0][c]

    # diagonals
    if b[0][0] == b[1][1] == b[2][2] != "-":
        return b[0][0]
    if b[0][2] == b[1][1] == b[2][0] != "-":
        return b[0][2]

    return None

 def is_draw(self):
    for r in range(3):
        for c in range(3):
            if self.board[r][c] == "-":
                return False
    return True
     
 def reset_board(self):
    self.board = [["-" for _ in range(3)] for _ in range(3)]

            
game = tictactoe()
game.print_board()

def play_game():
    game = tictactoe()

    while True:
        game.reset_board()
        print("New Game!")
        game.print_board()

        while True:
            # Human move
            game.make_human_move()
            game.print_board()

            winner = game.check_winner()
            if winner:
             if winner =="X":
              print("You have won!! :) ")
             elif winner =="O":
              print("Computer has won!! :) ")
             break

            if game.is_draw():
                print("It's a draw!")
                break

            # Computer move
            game.make_computer_move()
            game.print_board()

            winner = game.check_winner()
            if winner:
              if winner == "X":
                print("You have won!! :) ")
              elif winner =="O":
                print("Computer has won!! :) ")
              else: 
                print("error")
              break


            if game.is_draw():
                print("It's a draw!")
                break

        again = input("Play again? (y/n): ").lower()
        if again != "y":
            print("Byebye!")
            break

```

```python
play_game()
```
